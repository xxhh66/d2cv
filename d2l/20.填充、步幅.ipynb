```python
# 计算填充、步幅后的大小
import torch
from torch import nn

X  =  torch.rand(size=(8,8))
# print(f'X size:\n {X.shape},X:\n {X}')

# 经过卷积后X的大小
def comp_conv2d(conv2d,X):
    # X shape (1,1,8,8)
    X = X.reshape((1,1)+X.shape)
    print(f'X size:\n {X.shape}')
    Y = conv2d(X)
    return Y.reshape(Y.shape[2:])

# 核大小 3 填充 1
conv2d = nn.Conv2d(1,1,kernel_size=3,padding=1)
print(comp_conv2d(conv2d=conv2d,X=X).shape)
```

```python
import torch
import torch.nn as nn
import numpy as np

# ========== 1. 基础验证：1x1卷积的输入输出 ==========
def basic_1x1_conv_demo():
    """
    基础演示：1x1卷积核的输入输出关系
    """
    print("=" * 60)
    print("1. 基础演示：1x1卷积核的输入输出")
    print("=" * 60)
    
    # 输入：1个样本，3个通道，4x4的特征图
    x = torch.randn(1, 3, 4, 4)
    print(f"输入形状: {x.shape}  # (batch, channels, height, width)")
    
    # 1x1卷积层：输入3通道 → 输出2通道
    conv1x1 = nn.Conv2d(in_channels=3, out_channels=2, kernel_size=1, bias=False)
    
    # 前向传播
    y = conv1x1(x)
    print(f"输出形状: {y.shape}  # (batch, out_channels, height, width)")
    # print(f"\n输入通道数: 3 → 输出通道数: 2")
    # print(f"卷积核形状: {conv1x1.weight.shape}  # (out_channels, in_channels, 1, 1)")
    
    return x, conv1x1, y

basic_1x1_conv_demo()
```
