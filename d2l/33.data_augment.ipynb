```python
import os
import torch 
import torch.nn as nn

from torchvision import transforms
import matplotlib.pyplot as plt
from PIL import Image
```

```python
cat1 = Image.open("./datasets/cat1.jpg")

plt.imshow(cat1)
plt.show
```

```python
def apply(img, aug, num_rows=2, num_cols=4, figsize=(14, 7)):
    """
    应用数据增广并显示多张效果图
    
    Args:
        img: PIL Image 原始图片
        aug: torchvision.transforms 增广方法
        num_rows: 子图行数
        num_cols: 子图列数
        figsize: 图片大小 (宽, 高)
    
    Returns:
        显示增广效果图
    """
    print(f"transforms name:{aug.__class__.__name__}")
    total = num_rows * num_cols
    
    # 子图
    fig, axes = plt.subplots(num_rows, num_cols, figsize=figsize)
    axes = axes.flatten() if total > 1 else [axes]
    fig.suptitle(f"{aug.__class__.__name__}", fontsize=16, fontweight='bold')
    # 显示
    axes[0].imshow(img)
    axes[0].set_title("Original", fontsize=12)
    axes[0].axis('off')
    
    # Rest show augmented images
    for i in range(1, total):
        try:
            aug_img = aug(img)
            axes[i].imshow(aug_img)
            axes[i].set_title(f"Aug {i}", fontsize=12)
        except Exception as e:
            axes[i].imshow(img)
            axes[i].set_title(f"Aug {i}\n(Error)", fontsize=10)
        axes[i].axis('off')
    
    plt.tight_layout()
    plt.show()


# ============================================================
# 使用示例
# ============================================================

# 加载图片
img = Image.open("./datasets/cat1.jpg")

# 1. 测试 RandomVerticalFlip 垂直随机翻转增
aug = transforms.RandomVerticalFlip(p=0.5)
apply(img, aug, num_rows=2, num_cols=4)
```

```python
# 水平翻转
aug =transforms.RandomHorizontalFlip(p=0.5)
apply(img, aug, num_rows=2, num_cols=4)
```

```python
aug=transforms.RandomRotation(degrees=15)
apply(cat1,aug=aug,num_rows=2, num_cols=4)
```

```python
aug=transforms.RandomResizedCrop(size=(224, 224))
apply(cat1,aug=aug,num_rows=2, num_cols=4)
```

```python
aug=transforms.ColorJitter(brightness=0.5)
apply(cat1,aug=aug,num_rows=2, num_cols=4)
```

```python
aug=transforms.RandomAffine(degrees=20)
apply(cat1,aug=aug,num_rows=2, num_cols=4)
```

```python
aug=transforms.GaussianBlur(kernel_size=11)
apply(cat1,aug=aug,num_rows=2, num_cols=4)
```

```python
aug=transforms.RandomGrayscale(p=0.5)
apply(cat1,aug=aug,num_rows=2, num_cols=4)
```
