```python
import torch
from torch.utils.data import DataLoader
from torchvision import datasets,transforms
```

```python
def compute_mean_std(loader):
    """
    最常用的增量计算方式
    注：std 是近似值，但对于图像数据已足够精确
    """
    mean = 0.0
    std = 0.0
    total_samples = 0


    # 对数据loader进行for循环时， images: [batch_size, channels, height, width]， labels: [batch_size]
    for images, _ in loader:
        # [B,C,H,W]
        batch_samples = images.size(0)
        # 计算每个 batch 的均值和标准差
        batch_mean = images.mean(dim=[0, 2, 3])  # 计算当前批次每个通道的均值
        batch_std = images.std(dim=[0, 2, 3])    # 计算当前批次每个通道的方差
        
        # 加权累加（注意：std 的累加是近似值）
        mean += batch_mean * batch_samples
        std += batch_std * batch_samples
        
        total_samples += batch_samples
    
    mean /= total_samples
    std /= total_samples
    
    return mean, std

transform = transforms.Compose([
    # transforms.Resize((224, 224)),
    transforms.ToTensor(),
])
train_data = datasets.CIFAR10(root="./datasets",download=True,transform=transform)
train_loader = DataLoader(train_data, batch_size=64, shuffle=True, num_workers=4)

mean, std = compute_mean_std(train_loader)
print(f"Mean: {mean.tolist()}")
print(f"Std: {std.tolist()}")


# 格式化输出
mean_rounded = [round(m, 4) for m in mean.tolist()]
std_rounded = [round(s, 4) for s in std.tolist()]

print(f"Mean: {mean_rounded}")
print(f"Std:  {std_rounded}")
```
